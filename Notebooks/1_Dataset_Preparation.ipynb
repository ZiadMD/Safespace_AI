# 1. Dataset Preparation & Exploration

Safespace AI utilizes multiple datasets to test and train the models.
Use this notebook to explore your datasets, verify their Yolo formatting, and visualize random bounding boxes.


```python
import os
import glob
import random
import matplotlib.pyplot as plt
from PIL import Image

# Change this pointing to whichever dataset you want to test/explore
DATASET_PATH = '../Datasets/your_dataset_folder'
print(f'Using dataset at: {DATASET_PATH}')
```

```python
def check_yolo_dataset(dataset_path):
    images_train = os.path.join(dataset_path, 'images', 'train')
    labels_train = os.path.join(dataset_path, 'labels', 'train')
    
    if os.path.exists(images_train) and os.path.exists(labels_train):
        num_images = len(os.listdir(images_train))
        num_labels = len(os.listdir(labels_train))
        print(f'Train Images: {num_images} | Train Labels: {num_labels}')
    else:
        print('Dataset structure incomplete or paths incorrect.')

check_yolo_dataset(DATASET_PATH)
```
